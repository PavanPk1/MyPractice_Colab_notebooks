```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from scipy.stats import chi2
from scipy.stats import chi2_contingency
```

```python
data = pd.read_csv('adult_dataset.csv')
data.head()
```

```python
data.info()
```

```python
data.nunique()
```

```python
data.gender.value_counts()
```

```python
data.occupation.value_counts()
```

```python
print(data.groupby('occupation')['gender'].value_counts())
```

```python
pd.crosstab(data.occupation, data.gender, margins = True)
```

```python
observed = pd.crosstab(data.occupation, data.gender)
observed
```

```python
# chi2 test_statistic, p-value, degrees of freedom, expected_frequences
result = chi2_contingency(observed)
result
```

```python
chi2_test_stat = result[0]
deg_freedom = result[2]
expected_freq = result[3]

print(f"Chi-square test : {chi2_test_stat}")
print(f"Degrees of freedom: {deg_freedom}")
```

```python
#checking
print(2809*16192 / 48842)
print(5611*16192 / 48842)
print(2809*32650 / 48842)
print(5611*32650 / 48842)
```

```python
alpha = 0.05
chi2_critical = chi2.ppf(alpha, deg_freedom)
chi2_critical
```

```python
print(chi2_test_stat)
```

```python
if chi2_test_stat > chi2_critical:
    print('Reject Null Hypothesis')
else:
    print('Failed to reject Null Hypothesis')
```

### Effect of significance level and degree of freedom

```python
for i in [0.05, 0.1, 0.5, 0.99]: #significance level's
    for j in range(1,30,5): #degree of freedom
        print('For Significance Level : {} and Degree of Freedom : {} Chi2 Critical is = {}'.format(i,j,chi2.ppf(i,j)))
    print()
```

* we can conclude that with increase in significance level and degrees of freedom chi2 critical will also increase

* for a fixed df, increasing the significance level increases the critical value.
