```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
# Calculating Z score
def z_score(sample_size, sample_mean, pop_mean, pop_std):
    std_error = pop_std / sample_size ** 0.5
    z = (sample_mean - pop_mean) / std_error
    return z
```

## Q-1: A sports analyst claims that Virat Kohli has an average score of 40 in IPL matches with a standard deviation of 9. You doubt that the average score is **less than** 40 and take a sample of scores from 50 matches. Form the hypotheses and test the claim.


```python
sample = [41, 16, 57, 7, 33, 35, 36, 33, 43, 39, 38, 41, 35, 41, 39, 44, 39,
       38, 36, 43, 40, 39, 32, 40, 19, 36, 34, 33, 40, 89, 39, 36, 34, 44,
       43, 44, 41, 37, 40, 36, 68, 5, 37, 36, 34, 32, 40, 33, 34, 44]
```

**H0 :** The average score is 40. <br>
**H1 :** The average score is less than 40.

```python
sample = np.array(sample)
```

```python
pop_mean = 40
pop_std = 9
sample_size = len(sample)
sample_mean = sample.mean()
print('Sample Size : ', sample_size)
print('Sample Mean : ', sample_mean)
```

```python
# Calculate z score
z = z_score(sample_size, sample_mean, pop_mean, pop_std)
z
```

```python
# Lower tailed test (left tailed test)
# calculating p value
p_value = norm.cdf(z)
p_value
```

```python
# Conclusion
alpha = 0.05
if p_value < alpha:
    print('We have sufficient evidence to reject null hypothesis.')
else:
    print('We fail to reject null hypothesis')
```

## Q-2: A sports analyst claims that Virat Kohli has an average score of 40 in IPL matches with a standard deviation of 9. You doubt that the average score is more than 40 and take a sample of scores from 50 matches. Form the hypotheses and test the claim.

```python
sample = [41, 16, 57, 7, 33, 35, 76, 33, 43, 39, 18, 101, 35, 41, 39, 44, 39,
       38, 36, 43, 40, 3, 82, 40, 19, 36, 34, 33, 40, 89, 39, 36, 34, 44,
       43, 44, 11, 87, 40, 36, 68, 5, 37, 106, 34, 32, 40, 33, 34, 44]
```

**H0 :** The average score is 40. <br>
**H1 :** The average score is more than 40.

```python
sample = np.array(sample)
```

```python
pop_mean = 40
pop_std = 9
sample_size = len(sample)
sample_mean = sample.mean()
print('Sample Size : ', sample_size)
print('Sample Mean : ', sample_mean)
```

```python
# Calculate z score
z = z_score(sample_size, sample_mean, pop_mean, pop_std)
z
```

```python
# upper tailed test (Right tailed test)
# calculating p value
p_value = 1 - norm.cdf(z)
p_value
```

```python
# Conclusion
alpha = 0.05
if p_value < alpha:
    print('We have sufficient evidence to reject null hypothesis.')
else:
    print('We fail to reject null hypothesis')
```
