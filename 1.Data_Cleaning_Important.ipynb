## Duplicates

```python
# check which rows are duplicates (True/False)
df.duplicated()

# count duplicate rows
df.duplicated().sum()

# drop fully duplicate rows, keep first occurrence
df.drop_duplicates()

# drop duplicates based on specific columns
df.drop_duplicates(subset=["col1", "col2"])

# drop all duplicates (remove every row that has a duplicate)
df.drop_duplicates(keep=False)
```
