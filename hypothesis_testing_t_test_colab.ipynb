```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
from scipy.stats import t
```

## One Sample T test

```python
# Calculating t score (One sample test)
def t_score(sample_size, sample_mean, pop_mean, sample_std):
    std_error = sample_std / sample_size ** 0.5
    t = abs((sample_mean - pop_mean) / std_error)
    return t
```

**A company wants to test the claim that their batteries last more than 40 hours. Using a simple random sample of 30 batteries. Test this claim using a significance level of 0.05.**


```python
sample = [37, 36, 34, 36, 36, 39, 36, 43, 40, 34, 34, 35, 35, 35, 42, 40, 34,
       34, 36, 39, 37, 43, 38, 39, 43, 41, 39, 35, 43, 33]
```

**H0 :** The sample mean is 40. <br>
**H1 :** The sample mean is less than 40.

```python
sample = np.array(sample)
```

```python
sample.mean()
```

```python
pop_mean = 40
sample_size = len(sample)
sample_mean = sample.mean()
sample_std = sample.std()
df = sample_size - 1
print('Sample Size : ', sample_size)
print('Sample Mean : ', sample_mean)
print('Sample STD  : ', sample_std)
print('Degrees of freedom : ', df)
```

```python
# Calculate t score
t_value = t_score(sample_size, sample_mean, pop_mean, sample_std)
t_value
```

```python
# calculating critical t value
alpha = 0.05
t_critical = abs(t.ppf(alpha, df))
t_critical
```

```python
# Conclusion
if t_value > t_critical:
    print('We have sufficient evidence to reject null hypothesis.')
else:
    print('We fail to reject null hypothesis')
```

## Independent Samples T Test

```python
# Calculating t score (One sample test)
def t_score(sample1_size, sample1_mean, sample1_std, sample2_size, sample2_mean, sample2_std):

    temp = (sample1_std**2/sample1_size + sample2_std**2/sample2_size)**0.5

    t = abs((sample1_mean - sample2_mean) / temp)
    return t
```

```python
t_score(15, 84, 8, 15, 71, 9)
```

**A study is conducted to investigate whether the choice of exam preparation programs has an effect on exam scores (whether they are different).
For this study exam scores of a total of 60 students from two different programs(30 students each) were considered.**


```python
sample1 = [96, 71, 95, 85, 63, 92, 85, 97, 92, 83, 83, 69, 88, 93, 83, 96, 69,
       60, 71, 62, 64, 62, 98, 78, 92, 93, 77, 93, 79, 76]
```

```python
sample2 = [96, 80, 74, 62, 68, 73, 72, 65, 74, 52, 83, 98, 72, 96, 90, 91, 63,
       81, 89, 62, 57, 84, 67, 87, 69, 70, 83, 80, 78, 74]
```

**H0 :** Both the programs have equal mean scores. <br>
**H1 :** Both the programs have different mean scores.

```python
sample1 = np.array(sample1)
sample2 = np.array(sample2)
```

```python
sample1_size = len(sample1)
sample1_mean = sample1.mean()
sample1_std  = sample1.std()
sample2_size = len(sample2)
sample2_mean = sample2.mean()
sample2_std = sample2.std()

df = sample1_size + sample2_size - 2

print('Sample1 Size : ', sample1_size)
print('Sample1 Mean : ', sample1_mean)
print('Sample1 STD  : ', sample1_std)
print('Sample2 Size : ', sample2_size)
print('Sample2 Mean : ', sample2_mean)
print('Sample2 STD  : ', sample2_std)
print('Degrees of freedom : ', df)
```

```python
t_value = t_score(sample1_size, sample1_mean, sample1_std, sample2_size, sample2_mean, sample2_std)
t_value
```

```python
# Critical t value
alpha = 0.05
t_critical = abs(t.ppf(alpha, df))
t_critical
```

```python
# Conclusion
if t_value > t_critical:
    print('We have sufficient evidence to reject null hypothesis.')
else:
    print('We fail to reject null hypothesis')
```
